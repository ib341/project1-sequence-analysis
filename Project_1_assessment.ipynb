## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python

def validate_read(sequence, reference):
    """This function checks if a sequence is the same length as the reference and only contains valid characters (A/C/G/T)"""
    sequence = sequence.upper()  # makes sequence uppercase
    outcome = {
      "sequence": sequence,
      "valid": "",
      "errors": [],
      }
    if len(sequence) != len(reference):   # checks if the length matches the reference
      outcome["errors"].append("wrong_length")
    for base in sequence:   # checks if it contains only valid characters
      if base not in "ACGT":
        outcome["errors"].append("invalid_characters")
    if outcome["errors"] == []:   # decides if the sequence is valid or not
      outcome["valid"] = True
    else:
      outcome["valid"] = False
    return outcome    # gives the outcome

# add what to do if nothing inputed - type error
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """this function checks each base against the base in the reference string and records details about any mutations """
    mutations = []  # creates list to add mutations to
    for base in range(len(sequence)): # check each base against equal position reference base
      if sequence[base] != reference[base]: # if they don't match add to the list:
         mutations.append({
                    "position" : base +1, # the position of the mutation
                    "reference_base" : reference[base], # the reference base
                    "read_base" : sequence[base] # the mutated base
             })
    return mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:

    "sequence": "GTGA",             # normalised upper-case sequence

    "errors": [],                   # validation error codes

### 🔎 Expected outputs

- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """This function summerises each read to tell you various useful things including any errors and mutations using previously defined functions"""
    analysis = [] # list each entry wil be added to
    for read in reads: # for each dictionry in reads do the following
       seq = read["sequence"].upper() # makes it upper case
       read_analysis = { # creates the dictionary for the read
          "read_id" : read["read_id"],
          "sample" : read["sample"],
          "group" : read["group"],
          "sequence" : seq,
          "status" : "",
          "errors" : [],
          "mutations" : []
       }
       outcome = validate_read(seq, reference)
       if outcome["valid"] == True: # if the sequence is valaid, find the mutations
         read_analysis["status"] = "valid"
         read_analysis["mutations"] = find_mutations(seq, reference)
       else:
         read_analysis["status"] = "invalid" # if the sequence is not valid, say why
         read_analysis["errors"] = outcome["errors"]
       analysis.append(read_analysis)

    return analysis
````

### 💻 Enter your code here

````python
# Write your code here.
def summarise_samples(analysed_reads):
  """this function helps to summarise mismatch patterns between samples"""
  summary = {}
  def percent(x, y):  # function to make finding percent easier
    return summary[sample][x] / summary[sample][y] * 100


  for read in analysed_reads: # for each read record these names
    sample = read["sample"]
    group = read["group"]
    status = read["status"]
    if sample not in summary:  # if it is not already in the summary, create a sample dictionary with its name and add the following
      summary[sample] = {
        "group": group,
        "valid_reads": 0,
        "invalid_reads": 0,
        "mismatch_reads": 0,
        "mismatch_percentage": "None",
        "mismatch_events": 0,
        "substitution_counts": {},
        "position_counts": {},
        "top_positions": []
      }

    if status == "valid": # if a read is valid in the sample add one to valid reads
      summary[sample]["valid_reads"] += 1
    if status == "invalid": # if a read is invalid in the sample add one to invalid reads
      summary[sample]["invalid_reads"] += 1

    if "mutations" in read: # if mutations are present do the following
      if len(read["mutations"]) > 0:
        summary[sample]["mismatch_reads"] += 1 # add one to the mismatch reads counter
      for mutation in read["mutations"]:
        position = mutation["position"] # record the position of the mutation
        reference_base = mutation["reference_base"] # record the reference bas
        read_base = mutation["read_base"] # record the read base
        substitution = f"{reference_base}>{read_base}" # add a substation stating what changed
        summary[sample]["mismatch_events"] += 1 # add one to the mismatch events counter
        summary[sample]["substitution_counts"][substitution] = summary[sample]["substitution_counts"].get(substitution, 0) + 1 # if the substitution has occurred before it adds one to the count
        summary[sample]["position_counts"][position] = summary[sample]["position_counts"].get(position, 0) + 1 # if the position has been mutated before it adds one to the count

      if summary[sample]["valid_reads"] > 0: # if there is 1 or more valid reads
        summary[sample]["mismatch_percentage"] = percent("mismatch_events", "valid_reads") # creates the percentage of reads that have a mismatch
      if len(summary[sample]["position_counts"]) >= 3: # if there is more than 3 mutated positions
        sorted_positions = sorted(summary[sample]["position_counts"], key=summary[sample]["position_counts"].get, reverse=True) # rank the mutations by highest occurrence
        summary[sample]["top_positions"] = sorted_positions[:3] # keeps the top 3
      else:
        summary[sample]["top_positions"] = list(summary[sample]["position_counts"].keys())

  """table one"""

  groups = {}
  for sample, data in summary.items(): # sorts the data into groups
    group = data["group"]
    if group not in groups:
      groups[group] = []
    groups[group].append(sample)

  for group, samples in groups.items(): # per group make the following tables
    print(f'Group: {group}') # states which group the table is for
    print(f'{"Sample":<10}{"Valid reads":<15}{"Invalid reads":<15}{"Mismatch events percentage":<30}{"Total mismatch events":<25}{"Top three positions with highest mismatch events":<25}') # table 1
    print("-" * 144)
    for sample in samples:
     print(f'{sample:<10}{summary[sample]["valid_reads"]:<15}{summary[sample]["invalid_reads"]:<15}{summary[sample]["mismatch_percentage"]:<30}{summary[sample]["mismatch_events"]:<25}{str(summary[sample]["top_positions"]):<25}')

    print() # adds a empty line between the two tables
    """table two"""
    header = f'{"Substitution":<15}' # creates header
    for sample in samples:
      header += f'{sample:<20}' # adds each sample as a header
    print(header)
    print("-" * 150) # creates dashed line to separate from header

    all_substitutions = set()
    for sample in samples: # collects all the substitutions
      all_substitutions.update(summary[sample]["substitution_counts"].keys())
    for sub in sorted(all_substitutions):
      row_str = f"{sub:<15}"  # adds the substitution as a row
      for sample in samples: # finds the number of times its been substituted and number of valid reads
        count = summary[sample]["substitution_counts"].get(sub, 0)
        valid_reads = summary[sample]["valid_reads"]
        if valid_reads > 0: # if there is more than one valid read it generates a percentage
          percentage = (count / valid_reads) * 100
          cell_val = f"{count} ({percentage:.1f}%)"
        else: # if no valid reads it adds n/a
          cell_val = f"{count} (N/A)"
        row_str += f"{cell_val:<20}" # adds number of substations and percentages to the table
      print(row_str)
      print() # adds a empty line between the groups
  return summary



````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

```python
[
    {
        "position": 42,
        "control_n": 3,
        "treated_n": 3,
        "control_mean": 0.5,
        "treated_mean": 2.0,
        "difference": 1.5,
        "statistic": ...,
        "p_value": ...
    }
]
```

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# Write your code here.
def compare_positions(sample_summary, control_group="control", treated_group="treatment"):

    target_samples = {}
    for sample_id, sample_data in sample_summary.items(): # for each sample, check if it is either of the groups and then add it to a new dictionary
        group = sample_data.get("group")
        if group in [control_group, treated_group]:
            target_samples[sample_id] = sample_data

    combined_counts = {}
    for sample_data in target_samples.values():
        pos_counts = sample_data.get("position_counts", {}) # finds the positions
        for pos, count in pos_counts.items():
            combined_counts[pos] = combined_counts.get(pos, 0) + count

    sorted_positions = sorted(combined_counts.items(), key=lambda x: (-x[1], x[0])) # sorts the list from high to low, if theres a tie picks the earlier one

    top_positions = [pos for pos, count in sorted_positions if count > 0][:3] # picks the top 3 positions if there are more than 3

    results = []
    for pos in top_positions: # creates blank lists to add to
        control_percentages = []
        treated_percentages = []

        for sample_id, sample_data in target_samples.items(): # gets the group and valid reads for each sample
            group = sample_data.get("group")
            total_valid = sample_data.get("valid_reads", 0)
            if total_valid == 0: # if there are no valid reads it skips it to prevent zero division errors
              continue

            mismatch_count = sample_data.get("position_counts", {}).get(pos, 0) # finds the mismatch count
            percentage = mismatch_count / total_valid * 100 # calculates the mismatch percentage

            if group == control_group: # adds the mismatch percentage to the right group
                control_percentages.append(percentage)
            elif group == treated_group:
                treated_percentages.append(percentage)

        control_n = len(control_percentages) # how many control reads there are
        treated_n = len(treated_percentages) # how many treated reads there are
        control_mean = sum(control_percentages) / control_n # calculates control mean
        treated_mean = sum(treated_percentages) / treated_n # calculates treated mean
        difference = treated_mean - control_mean # calculates difference

        stat_result = calculate_statistics(control_percentages, treated_percentages) # uses the provided function to generate statistics
        statistic = stat_result.get("statistic")
        p_value = stat_result.get("p_value")

        results.append({ # adds the generated data to the results dictionary
            "position": pos,
            "control_n": control_n,
            "treated_n": treated_n,
            "control_mean": control_mean,
            "treated_mean": treated_mean,
            "difference": difference,
            "statistic": statistic,
            "p_value": p_value
        })

    for position in results: # prints a summary line for each position
      print(f'Position: {position["position"]}, Number of control samples: {position["control_n"]}, Control mean: {position["control_mean"]:.2f}, Number of treated samples {position["treated_n"]}, Treated mean:{position["treated_mean"]:.2f}, Difference of {position["difference"]:.2f}, p-value {position["p_value"]}')

    return results
````
